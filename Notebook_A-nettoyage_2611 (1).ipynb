### • Partie A1 – Cadrage


#### **1.	Enregistrement**
**Faut-il un numéro, une déclaration ou une licence pour louer ?**  
Oui, il faut un numéro auprès de la municipalité pour les locations à court terme depuis le 1er janvier 2021. Le permis doit être renouvelé tous les ans et son numéro listé dans l’offre de location sur Airbnb. Chaque location du bien doit être déclarée en ligne à la municipalité avant l’arrivée des locataires.  

#### **2.	Résidence principale**
**Peut-on louer sa résidence principale, et dans quelle limite ?**  
Oui, mais sous des conditions strictes : l'hôte doit impérativement résider à l'année à l'adresse qu'il souhaite mettre en location, ainsi que payer la taxe de séjour (12,5 %) et la TVA (21 % depuis le 1er janvier 2026). Ces deux taxes sont les plus élevées de l’Union Européenne pour la location de courte durée. Le nombre maximum de nuits par an est de 30, sauf dans certains quartiers (9 sur les 25) ou il est de 15.

Pour une chambre d’hôtes, qui n’est pas cumulable avec une location de la résidence principale dans sa totalité, il n’y a pas de limite de nombre de nuits par an, mais les autres restrictions listées ci-dessus s’appliquent également. La chambre doit faire une surface maximale de 61 m2, représentant au maximum 40% de la surface habitable du logement. Certains quartiers interdisent complètement les chambres d’hôtes, tandis que d’autres les autorisent dans la limite de quotas annuels sur les permis (voir carte détaillée listée dans les sources).

#### **3.	Résidence secondaire ou investissement**
**Est-ce autorisé ? Faut-il une autorisation particulière ?**  
Posséder une résidence secondaire à Amsterdam est autorisé dans la limite d’une par personne et dans des conditions très strictes, l’une d’entre elles interdisant de louer le bien à d’autres particuliers. Cette option n’est donc pas envisageable pour la location de courte durée.

#### **4.	Durée minimale**
**En dessous de combien de nuits parle-t-on de location courte durée au sens de la règle ?**  
Comme précisé plus haut, il y a un nombre maximal autorisé de nuits pour les locations courte durée – 15 ou 30 selon les quartiers, mais pas de durée minimum de séjour dans ce cadre.

#### **5.	Sanctions et évolutions**
**Quelles sanctions? Un durcissement est-il annoncé ou en discussion ?**  
Toute location non déclarée à l’avance ou effectuée sans permis est passible d’une amende de 1500 € pour un particulier et de 3000 € pour une entreprise. L’amende peut être plus élevée pour d’autres infractions, telles que le dépassement du seuil de 15 ou 30 nuits par an.

Le durcissement de ces sanctions et de la loi en général a déjà eu lieu, faisant suite à des études  montrant les lacunes de la législation européenne en matière de location de courte durée (Biber et al., 2017 ; von Briel & Dolnicar, 2021). Des études ultérieures ont montré une forte corrélation entre l’augmentation des locations à court terme à Amsterdam et celle des prix de l’immobilier, associée à une baisse de la qualité de vie dans les quartiers concernés (van Haaren et al., 2021 ; Valente et al., 2023). Les règles régissant le numéro de licence standardisé pour les locations de courte durée ont été implémentées à l’échelle de l’UE en avril 2024, suivies de restrictions additionnelles de la municipalité d’Amsterdam au 1er janvier 2026.

### • Partie A2 – Exploration et dictionnaire 

- **'listings' et 'reviews' : nombre de lignes et de colonnes, types de variables.**

```python
import pandas as pd

listings=pd.read_csv('listings.csv.gz')
reviews=pd.read_csv('reviews.csv.gz')

# Le nombre de lignes et de colonnes 
review_shape=reviews.shape
listings_shape=listings.shape

# Les types
review_dtype=reviews.dtypes.value_counts()
listings_dtype=listings.dtypes.value_counts()

# Résumé
def resume(df):
    infos = {'lignes': df.shape[0], 'colonnes': df.shape[1]}
    infos.update(df.dtypes.astype(str).value_counts().to_dict())   # nb de colonnes par type
    return infos

recap = pd.DataFrame({'Table reviews': resume(reviews), 'Table listings': resume(listings)}).fillna(0).astype(int)
display(recap)
```

- **Pourcentage de valeurs manquantes par colonne ('listings')**

```python
# % de valeurs manquantes par colonne, arrondi à 0,1, trié du plus grand au plus petit
valeurs_manquantesL=(listings.isna().mean() * 100).round(1).sort_values(ascending=False)

# Nombre de colonnes entièrement vides (100 % de NaN) et leur liste
n_empty=listings.isna().all().sum()
empty_cols=list(valeurs_manquantesL.index[valeurs_manquantesL==100])

# Regrouper par tranches de % et lister les colonnes de chaque tranche 
# pd.cut découpe les % en intervalles : ]-1;0], ]0;10], ]10;50], ]50;99,99], ]99,99;100]
groupes = pd.cut(valeurs_manquantesL,
                 bins=[-1, 0, 10, 50, 99.99, 100],
                 labels=['0 % (complètes)', '0–10 %', '10–50 %', '50–99 %', '100 % (vides)'])

# Table : % de manquants et tranche de chaque colonne
synthese = pd.DataFrame({'% manquants': valeurs_manquantesL, 'groupe': groupes})

# Par tranche : nombre de colonnes et noms des colonnes (séparés par des virgules)
recap = synthese.groupby('groupe', observed=False).agg(
    nb_colonnes=('% manquants', 'count'),
    colonnes=('% manquants', lambda s: ', '.join(s.index))
)

# Affichage sans couper le texte des cellules (liste des colonnes en entier)
with pd.option_context('display.max_colwidth', None):
    display(recap)
```

- **Pourcentage de valeurs manquantes par colonne ('reviews')**

```python
# % de valeurs manquantes par colonne, arrondi à 0,01, trié du plus grand au plus petit
valeurs_manquantesR = (reviews.isna().mean() * 100).round(2).sort_values(ascending=False)

# Nombre de colonnes entièrement vides (100 % de NaN)
n_empty = reviews.isna().all().sum()
print(f"{n_empty} colonne(s) entièrement vide(s) dans reviews\n")

# Tableau : une ligne par colonne avec son % de valeurs manquantes
display(valeurs_manquantesR.to_frame('% de valeurs manquantes'))
```

- **Nombre de valeurs distinctes**

```python
# Nombre de valeurs distinctes par colonne (NaN non comptées), triées du plus grand au plus petit
valeurs_dis_list = listings.nunique().sort_values(ascending=False)
valeurs_dis_rev  = reviews.nunique().sort_values(ascending=False)

# 'listings' : toutes les lignes affichées (max_rows = None, sinon pandas masque le milieu avec « … »)
print("Listings")
with pd.option_context('display.max_rows', None):
    display(valeurs_dis_list.to_frame('Valeurs distinctes'))

# 'reviews' : 6 colonnes seulement, pas besoin d'option d'affichage
print("Reviews")
display(valeurs_dis_rev.to_frame('Valeurs distinctes'))
```

```python
# Faire état des lieux-visualiser les problèmes, incoherences 
listings['price'].dtype

# Type de donneé objet (str=texte) alors que ça doit être float
listings['price'].dtype

# Visualiser les 20 valeurs random
listings['price'].sample(20)

# Voir tous les formats de prix présents dans la colonne 
formes = listings['price'].str.replace(r'\d', '9', regex=True)
formes.value_counts().head(15)

# Une fonctions pour convertir les prix dans le même format
def convertir_prix(x):
    if pd.isna(x):
        return None
    s = str(x).strip()
    s = s.replace('CAD', '').replace('dollars', '').replace('$', '').replace(' ', '')
    if ',' in s and '.' in s:         
        s = s.replace(',', '')
    elif ',' in s:
        apres_virgule = s.split(',')[1]       # après la virgule
        if len(apres_virgule) == 3:           # 3 ch → milliers  
            s = s.replace(',', '')            
        else:                                 # 2 ch → centimes 
            s = s.replace(',', '.')          
    try:
            return float(s)
    except ValueError:
            return None

new_table['price']=new_table['price'].apply(convertir_prix)

price_values1=listings['price'].isna().sum()
price_values2=new_table['price'].isna().sum()

summary_table={'avant':[price_values1,listings['price'].dtype ], 
               'apres':[price_values2,new_table['price'].dtype] }

# Conclusion
print("Convertir colonne 'price' en float et verifier que rien n'est perdu:")
pd.DataFrame(summary_table,index=['lignes', 'dtype'])
```

- **Signature**

```python
import hashlib
import pandas as pd

def signature(df, nom):
    """Résumé détaillé et empreinte d'une table nettoyée."""
    copie = df.copy()
    for col in copie.columns:
        # les listes (ex. amenities) sont converties en texte pour pouvoir être résumées
        if copie[col].map(lambda v: isinstance(v, (list, dict, set, tuple))).any():
            copie[col] = copie[col].map(str)
    empreinte = hashlib.md5(
        pd.util.hash_pandas_object(copie.astype(str), index=False).values.tobytes()
    ).hexdigest()

    print(f"=== Signature : {nom} ===")
    print(f"Lignes : {copie.shape[0]}          Colonnes : {copie.shape[1]}")
    print(f"Empreinte MD5 : {empreinte}")
    resume = pd.DataFrame({
        "type": copie.dtypes.astype(str),
        "manquants": copie.isna().sum(),
        "pct_manquants": (copie.isna().mean() * 100).round(2),
        "valeurs_distinctes": copie.nunique(),
    })
    display(resume)
    numeriques = copie.select_dtypes("number")
    if not numeriques.empty:
        display(numeriques.describe().T.round(2))
    return empreinte

signature(nouvelle_table_propre, "listings nettoyé")
signature(reviews_propre, "reviews nettoyé")
```
